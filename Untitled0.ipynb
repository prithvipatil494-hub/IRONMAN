```python
import random

class SnakeEnv:
    """Simple 400x400 Snake environment."""
    WIDTH = 400
    HEIGHT = 400
    DIRECTIONS = {
        "UP":    (0, -1),
        "DOWN":  (0,  1),
        "LEFT":  (-1,  0),
        "RIGHT": (1,  0),
    }
    DIR_LIST = list(DIRECTIONS.keys())

    def __init__(self):
        self.reset()

    def reset(self):
        # Start in the middle
        self.head_x = self.WIDTH // 2
        self.head_y = self.HEIGHT // 2
        self.direction = random.choice(self.DIR_LIST)
        # Initial body: 3 segments behind the head
        self.body = []
        # Place food randomly
        self._place_food()
        self.length = 5
        self.steps = 0
        self.done = False
        return self._get_state()

    def _place_food(self):
        # Random position not on the snake
        while True:
            x = random.randint(0, self.WIDTH - 1)
            y = random.randint(0, self.HEIGHT - 1)
            if (x, y) != (self.head_x, self.head_y) and (x, y) not in self.body:
                self.food_x = x
                self.food_y = y
                break

    def _get_danger(self):
        # Danger relative to current direction
        ahead_action = self.direction
        if self.direction == "UP":
            left_action = "LEFT"; right_action = "RIGHT"
        elif self.direction == "DOWN":
            left_action = "RIGHT"; right_action = "LEFT"
        elif self.direction == "LEFT":
            left_action = "DOWN"; right_action = "UP"
        elif self.direction == "RIGHT":
            left_action = "UP"; right_action = "DOWN"

        def is_danger(dx, dy):
            nx = self.head_x + dx
            ny = self.head_y + dy
            # Wall collision
            if nx < 0 or nx >= self.WIDTH or ny < 0 or ny >= self.HEIGHT:
                return True
            # Body collision (ignore tail tip if it will move)
            if (nx, ny) in self.body[:-1]:  # tail tip will move
                return True
            return False

        dr, dc = self.DIRECTIONS[self.direction]
        danger_ahead = is_danger(dr, dc)

        lr, lc = self.DIRECTIONS[left_action]
        danger_left = is_danger(lr, lc)

        rr, rc = self.DIRECTIONS[right_action]
        danger_right = is_danger(rr, rc)

        return danger_ahead, danger_left, danger_right

    def _get_state(self):
        danger_ahead, danger_left, danger_right = self._get_danger()
        # Build body list for state (list of [x,y] segments)
        body_list = [[x, y] for (x, y) in self.body]
        return {
            "head_x": self.head_x,
            "head_y": self.head_y,
            "food_x": self.food_x,
            "food_y": self.food_y,
            "direction": self.direction,
            "danger_ahead": danger_ahead,
            "danger_left": danger_left,
            "danger_right": danger_right,
            "snake_length": self.length,
            "body": body_list,
            "done": self.done,
        }

    def step(self, action):
        if self.done:
            return self._get_state(), 0, True, {}

        # Change direction if valid (cannot reverse)
        opposite = {
            "UP": "DOWN", "DOWN": "UP",
            "LEFT": "RIGHT", "RIGHT": "LEFT"
        }
        if action != opposite.get(self.direction):
            self.direction = action

        # Move head
        dx, dy = self.DIRECTIONS[self.direction]
        new_head_x = self.head_x + dx
        new_head_y = self.head_y + dy

        # Check wall collision
        if new_head_x < 0 or new_head_x >= self.WIDTH or new_head_y < 0 or new_head_y >= self.HEIGHT:
            self.done = True
            return self._get_state(), -500, True, {}

        # Check body collision (excluding tail if it will shrink)
        if (new_head_x, new_head_y) in self.body[:-1]:
            self.done = True
            return self._get_state(), -500, True, {}

        # Move body
        self.body.insert(0, (self.head_x, self.head_y))
        self.head_x = new_head_x
        self.head_y = new_head_y

        # Check food
        reward = 1
        if self.head_x == self.food_x and self.head_y == self.food_y:
            reward = 10
            self.length += 1
            self._place_food()
        else:
            # Remove tail if no food eaten
            if self.body:
                self.body.pop()

        self.steps += 1
        # Optional: max steps
        if self.steps > 2000:
            self.done = True

        return self._get_state(), reward, self.done, {}
```

```python
def state_to_prompt(state):
    dir = state["direction"]
    head_x, head_y = state["head_x"], state["head_y"]
    food_x, food_y = state["food_x"], state["food_y"]

    # Relative food position
    dx = food_x - head_x
    dy = food_y - head_y

    food_parts = []
    if dx > 0: food_parts.append(f"{dx} units to the RIGHT")
    elif dx < 0: food_parts.append(f"{-dx} units to the LEFT")
    if dy > 0: food_parts.append(f"{dy} units BELOW")
    elif dy < 0: food_parts.append(f"{-dy} units ABOVE")

    food_desc = " and ".join(food_parts) if food_parts else "directly on the food"

    # Danger mapping
    dir = state["direction"]
    if dir == "UP":
        ahead = "UP"; left = "LEFT"; right = "RIGHT"
    elif dir == "DOWN":
        ahead = "DOWN"; left = "RIGHT"; right = "LEFT"
    elif dir == "LEFT":
        ahead = "LEFT"; left = "DOWN"; right = "UP"
    elif dir == "RIGHT":
        ahead = "RIGHT"; left = "UP"; right = "DOWN"

    danger_ahead = "yes" if state["danger_ahead"] else "no"
    danger_left  = "yes" if state["danger_left"] else "no"
    danger_right = "yes" if state["danger_right"] else "no"

    prompt = (
        f"You are playing Snake on a 400x400 grid (x from 0 to 399, y from 0 to 399).\n"
        f"Your current direction: {dir}.\n"
        f"Food is {food_desc}.\n"
        f"Direct danger:\n"
        f"- Ahead ({ahead}): {danger_ahead}\n"
        f"- Left ({left}): {danger_left}\n"
        f"- Right ({right}): {danger_right}\n"
        f"Your current length: {state['snake_length']}.\n"
        f"Choose your next move (UP, DOWN, LEFT, RIGHT):"
    )
    return prompt
```

```python
import os, time, random, warnings
import torch
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.animation import FuncAnimation
from IPython import display

# Suppress the max_new_tokens / max_length warning
warnings.filterwarnings("ignore", message="Both `max_new_tokens`.*and `max_length`")
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from unsloth import FastLanguageModel
from peft import PeftModel

# ─────────────────────────── Config ────────────────────────────
GRID_W, GRID_H = 30, 30          # Logical grid (display cells)
CELL_PX       = 20               # Pixels per cell in the figure
FIG_W         = GRID_W * CELL_PX / 100
FIG_H         = GRID_H * CELL_PX / 100 + 0.6   # +0.6 for title bar
MAX_STEPS     = 2000

# Colors
COLOR_BG      = "#1a1a2e"
COLOR_HEAD    = "#00ff88"
COLOR_BODY    = "#00aa55"
COLOR_FOOD    = "#ff4444"
COLOR_BORDER  = "#e94560"
COLOR_TEXT    = "#eaeaea"
COLOR_SCORE   = "#f5a623"

# ─────────────────────────── Environment ───────────────────────
class SnakeEnv:
    DIRECTIONS = {"UP":(0,-1), "DOWN":(0,1), "LEFT":(-1,0), "RIGHT":(1,0)}
    DIR_LIST   = list(DIRECTIONS.keys())

    def __init__(self, width=GRID_W, height=GRID_H):
        self.W, self.H = width, height
        self.reset()

    def reset(self):
        self.head_x   = self.W // 2
        self.head_y   = self.H // 2
        self.direction = random.choice(self.DIR_LIST)
        self.length    = 5
        self.body      = []
        opp = {"UP":"DOWN","DOWN":"UP","LEFT":"RIGHT","RIGHT":"LEFT"}
        dx, dy = self.DIRECTIONS[opp[self.direction]]
        for i in range(1, self.length):
            self.body.append((self.head_x + dx*i, self.head_y + dy*i))
        self._place_food()
        self.steps = 0
        self.score = 0
        self.done  = False
        return self._get_state()

    def _place_food(self):
        occupied = set(self.body) | {(self.head_x, self.head_y)}
        while True:
            x = random.randint(0, self.W-1)
            y = random.randint(0, self.H-1)
            if (x, y) not in occupied:
                self.food_x, self.food_y = x, y
                break

    def _get_danger(self):
        d = self.direction
        turns = {"UP":("LEFT","RIGHT"),"DOWN":("RIGHT","LEFT"),
                 "LEFT":("DOWN","UP"),"RIGHT":("UP","DOWN")}
        L, R = turns[d]
        def danger(dx, dy):
            nx, ny = self.head_x+dx, self.head_y+dy
            return (nx<0 or nx>=self.W or ny<0 or ny>=self.H
                    or (nx,ny) in self.body[:-1])
        return (danger(*self.DIRECTIONS[d]),
                danger(*self.DIRECTIONS[L]),
                danger(*self.DIRECTIONS[R]))

    def _get_state(self):
        da, dl, dr = self._get_danger()
        return {"head_x":self.head_x, "head_y":self.head_y,
                "food_x":self.food_x, "food_y":self.food_y,
                "direction":self.direction,
                "danger_ahead":da, "danger_left":dl, "danger_right":dr,
                "snake_length":self.length,
                "body":list(self.body), "done":self.done}

    def step(self, action):
        if self.done:
            return self._get_state(), 0, True, {}
        opp = {"UP":"DOWN","DOWN":"UP","LEFT":"RIGHT","RIGHT":"LEFT"}
        if action != opp.get(self.direction):
            self.direction = action
        dx, dy = self.DIRECTIONS[self.direction]
        nx, ny = self.head_x+dx, self.head_y+dy
        if nx<0 or nx>=self.W or ny<0 or ny>=self.H:
            self.done = True
            return self._get_state(), -500, True, {}
        if (nx, ny) in self.body[:-1]:
            self.done = True
            return self._get_state(), -500, True, {}
        self.body.insert(0, (self.head_x, self.head_y))
        self.head_x, self.head_y = nx, ny
        if self.head_x == self.food_x and self.head_y == self.food_y:
            reward = 10
            self.length += 1
            self.score  += 10
            self._place_food()
        else:
            reward = 1
            if self.body:
                self.body.pop()
        self.steps += 1
        if self.steps >= MAX_STEPS:
            self.done = True
        return self._get_state(), reward, self.done, {}


# ─────────────────────────── Prompt ────────────────────────────
def state_to_prompt(state):
    d  = state["direction"]
    dx = state["food_x"] - state["head_x"]
    dy = state["food_y"] - state["head_y"]
    parts = []
    if dx > 0:  parts.append(f"{dx} units to the RIGHT")
    elif dx < 0: parts.append(f"{-dx} units to the LEFT")
    if dy > 0:  parts.append(f"{dy} units BELOW")
    elif dy < 0: parts.append(f"{-dy} units ABOVE")
    food_desc = " and ".join(parts) if parts else "directly on the food"
    turns = {"UP":("UP","LEFT","RIGHT"),"DOWN":("DOWN","RIGHT","LEFT"),
             "LEFT":("LEFT","DOWN","UP"),"RIGHT":("RIGHT","UP","DOWN")}
    a, l, r = turns[d]
    da = "YES ⚠" if state["danger_ahead"] else "no"
    dl = "YES ⚠" if state["danger_left"]  else "no"
    dr = "YES ⚠" if state["danger_right"] else "no"
    return (f"You are playing Snake on a {GRID_W}x{GRID_H} grid.\n"
            f"Current direction: {d}. Food is {food_desc}.\n"
            f"Danger — Ahead ({a}): {da} | Left ({l}): {dl} | Right ({r}): {dr}\n"
            f"Length: {state['snake_length']}.\n"
            f"Choose your next move (UP, DOWN, LEFT, RIGHT):")


# ─────────────────────────── Model ─────────────────────────────
print("🔧 Loading model …")
base_model = "unsloth/Qwen2.5-1.5B-Instruct-bnb-4bit"
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=base_model, max_seq_length=512, load_in_4bit=True
)
model = PeftModel.from_pretrained(model, "outputs/checkpoint-500",
                                   local_files_only=True)
FastLanguageModel.for_inference(model)
print("✅ Model ready.\n")


def get_action(state):
    prompt  = state_to_prompt(state)
    inputs  = tokenizer(prompt, return_tensors="pt").to(model.device)
    in_len  = inputs["input_ids"].shape[1]
    # Suppress the specific HuggingFace warning at call time too
    with torch.no_grad(), warnings.catch_warnings():
        warnings.simplefilter("ignore")
        outputs = model.generate(
            **inputs,
            max_new_tokens=5,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )
    text = tokenizer.decode(outputs[0][in_len:],
                            skip_special_tokens=True).strip().upper()
    for d in ("UP", "DOWN", "LEFT", "RIGHT"):
        if d in text:
            return d
    return random.choice(["UP", "DOWN", "LEFT", "RIGHT"])


# ─────────────────────────── Renderer ──────────────────────────
def render(env, ax, reward=0, last_action=""):
    ax.clear()
    ax.set_facecolor(COLOR_BG)
    ax.set_xlim(0, env.W)
    ax.set_ylim(0, env.H)
    ax.set_aspect("equal")
    ax.axis("off")

    # Border
    border = patches.FancyBboxPatch(
        (0, 0), env.W, env.H,
        boxstyle="round,pad=0.1",
        linewidth=2, edgecolor=COLOR_BORDER, facecolor="none"
    )
    ax.add_patch(border)

    # Food  (pulsing star-like marker)
    ax.plot(env.food_x + 0.5, env.H - env.food_y - 0.5,
            marker="*", markersize=14, color=COLOR_FOOD, zorder=3)

    # Body
    body_set = set(env.body)
    for (bx, by) in env.body:
        rect = patches.FancyBboxPatch(
            (bx + 0.08, env.H - by - 1 + 0.08),
            0.84, 0.84,
            boxstyle="round,pad=0.05",
            linewidth=0, facecolor=COLOR_BODY, zorder=2
        )
        ax.add_patch(rect)

    # Head (slightly larger, brighter)
    head_rect = patches.FancyBboxPatch(
        (env.head_x + 0.04, env.H - env.head_y - 1 + 0.04),
        0.92, 0.92,
        boxstyle="round,pad=0.08",
        linewidth=1.5, edgecolor="#ffffff", facecolor=COLOR_HEAD, zorder=4
    )
    ax.add_patch(head_rect)

    # Eyes based on direction
    ex, ey_base = env.head_x + 0.5, env.H - env.head_y - 0.5
    offsets = {"UP":   [(-0.18, 0.22), (0.18, 0.22)],
               "DOWN": [(-0.18,-0.22), (0.18,-0.22)],
               "LEFT": [(-0.22,-0.12),(-0.22, 0.12)],
               "RIGHT":[(0.22,-0.12), (0.22, 0.12)]}
    for ox, oy in offsets.get(env.direction, [(0,0)]):
        ax.plot(ex+ox, ey_base+oy, "o",
                markersize=3, color="#000000", zorder=5)

    # HUD
    ax.set_title(
        f"🐍 Score: {env.score:>4}   Length: {env.length:>3}   "
        f"Steps: {env.steps:>5}   Move: {last_action:<6}   Reward: {reward:+}",
        color=COLOR_TEXT, fontsize=9, pad=4,
        fontfamily="monospace",
        bbox=dict(facecolor="#0f0f1a", edgecolor=COLOR_BORDER,
                  boxstyle="round,pad=0.3")
    )


# ─────────────────────────── Play loop ─────────────────────────
env   = SnakeEnv()
state = env.reset()

fig, ax = plt.subplots(figsize=(FIG_W, FIG_H))
fig.patch.set_facecolor(COLOR_BG)
plt.tight_layout(pad=0.4)

total_reward = 0
last_action  = env.direction
step_reward  = 0

print("🎮 AI Snake is playing …  (Ctrl-C to stop)\n")

try:
    while not state["done"]:
        action      = get_action(state)
        state, step_reward, done, _ = env.step(action)
        total_reward += step_reward
        last_action   = action

        render(env, ax, reward=step_reward, last_action=action)
        display.clear_output(wait=True)
        display.display(fig)
        plt.pause(0.001)

        if done:
            break

except KeyboardInterrupt:
    print("\n⏹  Interrupted by user.")

# Final frame
render(env, ax, reward=step_reward, last_action=last_action)
display.clear_output(wait=True)
display.display(fig)

print(f"\n{'='*50}")
print(f"🏁  Game Over!")
print(f"    Final score  : {env.score}")
print(f"    Snake length : {env.length}")
print(f"    Total steps  : {env.steps}")
print(f"    Total reward : {total_reward}")
print(f"{'='*50}")
```
